# Etapa 1: Inspeção Inicial dos Dados (Kaggle House Prices)

Este notebook realiza o carregamento inicial dos dados de treino e teste, inspeciona as dimensões da tabela, os tipos de variáveis e analisa estatisticamente a nossa variável alvo (**SalePrice**).

```python
import pandas as pd

# 1. Carregando os dados
df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

print(f"-> Treino : {df_train.shape[0]} imóveis e {df_train.shape[1]} colunas")
print(f"-> Teste  : {df_test.shape[0]} imóveis e {df_test.shape[1]} colunas")

# Exibindo as 3 primeiras linhas do conjunto de treino
df_train.head(3)
```

### 1.1 Análise das Dimensões e da Coluna Alvo
- O conjunto de treino possui **1.460 imóveis** e **81 colunas**.
- O conjunto de teste possui **1.459 imóveis** e **80 colunas**.
- A única coluna que está no treino e não está no teste é a **`SalePrice`** (o preço de venda), exatamente o valor que o modelo precisa prever.

```python
preco = df_train["SalePrice"]

print("=" * 45)
print(" RESUMO FINANCEIRO DOS PREÇOS DE VENDA")
print("=" * 45)
print(f"Menor Preço : US$ {preco.min():>12,.2f}")
print(f"Preço Médio : US$ {preco.mean():>12,.2f}")
print(f"Mediana     : US$ {preco.median():>12,.2f}")
print(f"Maior Preço : US$ {preco.max():>12,.2f}")
print("=" * 45)

# Resumo estatístico detalhado
preco.describe().to_frame()
```

### 1.2 Interpretação dos Preços de Venda
- O imóvel mais barato foi vendido por **US\$ 34.900,00** e o mais caro por **US\$ 755.000,00**.
- A **média** (US\$ 180.921,20) é superior à **mediana** (US\$ 163.000,00).
- **Conceito de Machine Learning:** Isso demonstra que a distribuição de preços tem **assimetria à direita** (*right-skewed*), pois poucas mansões muito caras elevam a média. Por isso, mais adiante aplicaremos uma transformação logarítmica (`np.log1p`) para deixar a distribuição simétrica e facilitar o aprendizado dos modelos.

```python
# Separando colunas numéricas e categóricas
colunas_numericas = df_train.select_dtypes(include=["number"]).columns.drop(["Id", "SalePrice"])
colunas_categoricas = df_train.select_dtypes(include=["object", "string"]).columns

print(f"Total de variáveis preditoras: {len(colunas_numericas) + len(colunas_categoricas)}")
print(f"  -> Numéricas   (metragens, anos, contagens) : {len(colunas_numericas)}")
print(f"  -> Categóricas (textos, siglas, categorias) : {len(colunas_categoricas)}")
```

### 1.3 Conclusão da Inspeção Inicial
Temos uma visão clara do formato dos dados:
1. **1.460 amostras de treino** com **79 variáveis preditoras**.
2. **36 variáveis numéricas** (que poderão ser normalizadas/escaladas).
3. **43 variáveis categóricas** (que precisarão ser convertidas em números via *Ordinal Encoding* ou *One-Hot Encoding*).
